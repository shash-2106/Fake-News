```python
from indicnlp.tokenize import indic_tokenize
from indicnlp.syllable import syllabifier
```

```python
import re
def preprocess_text(text):
  text = re.sub(r'd+','',text)
  return indic_tokenize.trivial_tokenize(text,'ml')

```

```python
df['tokenized_news'] = df['News/Comment'].apply(preprocess_text)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
```

```python
df['sentences_as_str'] = df['tokenized_news'].apply(lambda tokens: ' '.join(tokens))
```

```python
x = vectorizer.fit_transform(df['sentences_as_str'])
```

```python
from sklearn import preprocessing
import numpy as np
```

##Labelling

```python
label = {'FALSE':0,
         'HALF TRUE':1,
         'MOSTLY FALSE':2,
         'PARTLY FALSE':3,
         'MOSTLY TRUE':4}
```

```python
y = df['Type'].map(label)
```

##Train test splitting

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=100)
```

```python
from sklearn import svm
sl = svm.SVC(kernel='linear')
sl.fit(x_train,y_train)
```

##Applying model to make predction

```python
y_sl_train_pred = sl.predict(x_train)
y_sl_test_pred = sl.predict(x_test)
```

##Testing accuracy

```python
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score
```

```python
print(classification_report(y_test, y_sl_test_pred))
```
