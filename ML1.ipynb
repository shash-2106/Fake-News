#*ML Challenge*
##Loading data

```python
import pandas as pd
import numpy as np
from google.colab import drive
drive.mount('/content/drive')
```

```python
df = pd.read_csv('/content/drive/MyDrive/ML/labelled_train_set.csv')
```

```python
import nltk
nltk.download('punkt')
from nltk.tokenize import word_tokenize
```

```python
df['News/Comment'] = df['News/Comment'].fillna('')
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()
```

```python
x = vectorizer.fit_transform(df['News/Comment'])
```

##Labelling

```python
label = {'FALSE':0,
         'HALF TRUE':1,
         'MOSTLY FALSE':2,
         'PARTLY FALSE':3,
         'MOSTLY TRUE':4}
```

```python
y = df['Type'].map(label)
```

##Train test splitting

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=100)
```

```python
from sklearn.linear_model import LogisticRegression
lg = LogisticRegression()
lg.fit(x_train,y_train)
```

##Applying model to make predction

```python
y_lg_train_pred = lg.predict(x_train)
y_lg_test_pred = lg.predict(x_test)
```

##Testing accuracy

```python
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score
```

```python
print(classification_report(y_train, y_lg_train_pred))
```
